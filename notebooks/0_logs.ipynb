### 💯 Basic concepts of wandb

- `Experiment tracking` after all
- 
- `Run`: One experiment = one training attempt with a specific setup
- `Config`: What I tried
  - Hyperparameters
  - Model type
  - Dataset version
- `Logs`: What actually happened
  - Result of training
  - Traning loss
  - Validation accuracy
  - Metrics overtime
- `Dashboard`: wandb gives you to compare result

### ❌ Don't run this, this is a template for `wandb` workflow

```python
import random

import wandb

run = wandb.init(
    entity="trungdunglebui17112004-ho-chi-minh-city-university-of-te",
    project="sentiment-analysis-on-movie-reviews",
    # 1️⃣ Set the configuration of this run
    config={
        "learning_rate": 0.02,
        "architecture": "CNN",
        "dataset": "CIFAR-100",
        "epochs": 10,
    },
)

# Experiment by training with that configuration
epochs = 10
offset = random.random() / 5
for epoch in range(2, epochs):
    acc = 1 - 2**-epoch - random.random() / epoch - offset
    loss = 2**-epoch + random.random() / epoch + offset

    # 2️⃣ Log metrics to wandb
    run.log({"acc": acc, "loss": loss})

# 3️⃣ Log the run
run.finish()
```
